```python
from pyspark.sql import functions as F

data = [
    (1, "Laptop", 75000),
    (2, "Mobile", 25000),
    (3, "Headphones", 3000),
    (4, "Keyboard", 2000)
]

columns = [
    "product_id",
    "product_name",
    "price"
]

df = spark.createDataFrame(data, columns)

display(df)
```

```python
result = df.withColumn(
    "price_with_tax",
    F.round(F.col("price") * 1.18, 2)
)

display(result)
```

```python
df.createOrReplaceTempView("sample_products")

result = spark.sql("""
    SELECT
        COUNT(*) AS total_products,
        SUM(price) AS total_price,
        AVG(price) AS average_price
    FROM sample_products
""")

display(result)
```
